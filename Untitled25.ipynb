###**The pipeline**

\- Load CelebA dataset, select smile labels.

\- Build Conditional Variational Autoencoder (CVAE):

    - Encoder: compress image + label → latent vector

    - Decoder: reconstruct image from latent + label

    - Train VAE on training data.

    - Generate smiling images from non-smiling faces.

    - Visualize original and generated images.

```python
# Core libraries
import torch
import torchvision
from torch import nn, optim
from torch.utils.data import DataLoader, random_split
from torchvision import transforms, datasets
import matplotlib.pyplot as plt
import numpy as np
```

###**Define transformations for the dataset: resize images to 64x64 and convert them to tensors**

```python
# Define transformations for the dataset: resize images to 64x64 and convert them to tensors
transform = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.ToTensor() #converts images from PIL format to PyTorch tensors :values between 0 and 1
])
```

###**Download and load the CelebA dataset (faces dataset)**


CelebA contains 200k+ celebrity images with labels like smiling

```python

dataset = torchvision.datasets.CelebA(
    root="./data", # folder to save dataset
    split="train", # use training split
    target_type="attr", # load facial attributes
    transform=transform,  # apply the transformation defined above
    download=True  # download if not already present
)
```

###**Function to get "smiling" label from attributes**

```python

def get_smile_label(attr):
    # index 31 corresponds to 'Smiling' attribute
    #  only care whether the person is smiling (1) or not (0)
    return 1 if attr[31] == 1 else 0
```

###**Prepare a smaller dataset (15000 images) to speed up training**


\- Collects 15k images and their smile labels.

\- Reducing dataset for faster computation.

```python
images = []
labels = []

for i in range(15000):  # dataset réduit
    img, attr = dataset[i] # get image and attributes
    images.append(img) # store image tensor
    labels.append(get_smile_label(attr)) # store smile label
```

```python
#  Convert lists to tensors
images = torch.stack(images)  # shape: [15000, 3, 64, 64]
labels = torch.tensor(labels)  # shape: [15000]

# Create a PyTorch dataset combining images and labels
full_dataset = torch.utils.data.TensorDataset(images, labels)
```

```python
# Split dataset into train (80%) and test (20%)
train_size = int(0.8 * len(full_dataset))
test_size = len(full_dataset) - train_size

train_ds, test_ds = random_split(full_dataset, [train_size, test_size])

# DataLoaders for batch processing
train_loader = DataLoader(train_ds, batch_size=128, shuffle=True)
test_loader = DataLoader(test_ds, batch_size=128, shuffle=False)
```

###**Model parameters**

```python
# size of latent vector (compressed representation)
latent_dim = 128
# number of labels: smiling / not smiling
num_classes = 2
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(device)
```

###**Endoer**

\- Encoder compresses image + label into a vector.

\- Conv2d layers extract features.

\- 3 + 1 because we concatenate label as an extra channel.

\- Flatten converts feature maps to a single vector.

```python
encoder = nn.Sequential(
    nn.Conv2d(3 + 1, 32, 4, 2, 1), # input: 3 channels + 1 label channel
    nn.ReLU(),
    nn.Conv2d(32, 64, 4, 2, 1),
    nn.ReLU(),
    nn.Conv2d(64, 128, 4, 2, 1),
    nn.ReLU(),
    nn.Flatten()   # flatten to 1D vector
).to(device)
```

###**Linear layers to predict mean and log variance of latent distribution**

In VAE, latent variables are sampled from Gaussian with mean (mu) and variance (logvar).

```python
fc_mu = nn.Linear(128 * 8 * 8, latent_dim).to(device)
fc_logvar = nn.Linear(128 * 8 * 8, latent_dim).to(device)
```

###**Decoder**

\- Decoder input includes latent vector plus class label.

\- Map to same shape as flattened encoder output.

\- Decoder reconstructs images from latent vector + class.

\- ConvTranspose2d upsamples the feature maps.

\- Sigmoid ensures pixel values are between 0–1.

```python
decoder_input = nn.Linear(latent_dim + num_classes, 128 * 8 * 8).to(device)

decoder = nn.Sequential(
    nn.Unflatten(1, (128, 8, 8)), # reshape to feature maps
    nn.ConvTranspose2d(128, 64, 4, 2, 1),
    nn.ReLU(),
    nn.ConvTranspose2d(64, 32, 4, 2, 1),
    nn.ReLU(),
    nn.ConvTranspose2d(32, 3, 4, 2, 1),
    nn.Sigmoid()  # output image pixels between 0 and 1
).to(device)
```

###**Reparameterization & loss**

```python
def reparameterize(mu, logvar):
    std = torch.exp(0.5 * logvar) # standard deviation
    eps = torch.randn_like(std) # random noise
    return mu + eps * std # reparameterization trick

def vae_loss(recon_x, x, mu, logvar):
    recon_loss = nn.functional.mse_loss(recon_x, x, reduction="sum") # reconstruction loss
    kl = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())  # KL divergence
    return recon_loss + kl
```

###**Optimizer & training loop**

\- Combine all model parameters for the optimizer.

\- Adam optimizer with learning rate 0.001.

```python
params = list(encoder.parameters()) + \
         list(fc_mu.parameters()) + \
         list(fc_logvar.parameters()) + \
         list(decoder.parameters()) + \
         list(decoder_input.parameters())
```

\- Training loop for 20 epochs.

Each batch:

\- Condition image on smile label.

\- Encode → latent vector → sample z.

\- Decode z → reconstructed image.

\- Compute VAE loss.

\- Backpropagate and update weights.

\- Print average loss per epoch.

```python
optimizer = optim.Adam(params, lr=1e-3)

epochs = 20

for epoch in range(epochs):
    encoder.train() # set model to training mode
    total_loss = 0

    for x, y in train_loader:
        x = x.to(device)
        y = y.to(device)

        # create label channel for conditioning
        y_img = y.view(-1, 1, 1, 1).repeat(1, 1, 64, 64)
        x_cond = torch.cat([x, y_img], dim=1) # concatenate label channel

        h = encoder(x_cond)  # encode image
        mu, logvar = fc_mu(h), fc_logvar(h) # get mean and logvar
        z = reparameterize(mu, logvar)  # sample latent vector

        y_onehot = nn.functional.one_hot(y, num_classes=2).float()
        z = torch.cat([z, y_onehot], dim=1) # concatenate class info

        recon = decoder(decoder_input(z)) # decode to image

        loss = vae_loss(recon, x, mu, logvar)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    print(f"Epoch {epoch+1} | Loss: {total_loss/len(train_loader):.2f}")

```

###**Evaluation / generating smiling faces**

\- Take non-smiling images, force label to smiling.

\- Encode -> combine latent + label, decode ->  generate smiling faces.

\- Plot original non-smiling faces and generated smiling faces below them.

```python
# switch to evaluation mode
encoder.eval()
decoder.eval()

# fetch non-smiling images from test set
while True:
    x, y = next(iter(test_loader))
    mask = (y == 0)  # non-smiling
    if mask.sum() > 0:
        x = x[mask]
        y = y[mask]
        break

x = x.to(device)
y = y.to(device)

# limit to 10 images
n = min(10, x.size(0))
x = x[:n]

# target condition: smiling
target_smile = torch.ones(n, dtype=torch.long).to(device)

# condition image (for encoder)
y_img = target_smile.view(-1,1,1,1).repeat(1,1,64,64)
x_cond = torch.cat([x, y_img], dim=1)

with torch.no_grad():
    h = encoder(x_cond)
    mu, logvar = fc_mu(h), fc_logvar(h)
    z = mu  # deterministic at test time

    y_onehot = nn.functional.one_hot(target_smile, 2).float()
    z = torch.cat([z, y_onehot], dim=1)

    out = decoder(decoder_input(z))

```

```python
plt.figure(figsize=(20,5))  # wider figure

for i in range(n):
    # original (non-smiling)
    plt.subplot(2, n, i+1)
    plt.imshow(x[i].cpu().permute(1,2,0))
    plt.title("Original\n(No Smile)", fontsize=9)
    plt.axis("off")

    # generated (smiling)
    plt.subplot(2, n, i+1+n)
    plt.imshow(out[i].cpu().permute(1,2,0))
    plt.title("Generated\n(Smile)", fontsize=9)
    plt.axis("off")

plt.tight_layout(pad=1.5)
plt.show()
```
